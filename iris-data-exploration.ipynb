```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the built-in Iris dataset directly from seaborn
iris_df = sns.load_dataset('iris')

# Display dataset structure
print("--- Dataset Shape (Rows, Columns) ---")
print(iris_df.shape)

print("\n--- Column Names ---")
print(iris_df.columns)

print("\n--- First 5 Rows of Data ---")
print(iris_df.head())
```

```python
# ---------------------------------------------------------------------
# DATA SUMMARY & STATISTICS
# ----------------------------------------------------------------------

print("--- Missing Values Audit ---")
print(iris_df.isnull().sum())

print("\n--- Statistical Summary ---")
print(iris_df.describe())
```

```python
# ---------------------------------------------------------------------
# EXPLORATORY DATA ANALYSIS (VISUALIZATIONS)
# ----------------------------------------------------------------------
sns.set_theme(style="whitegrid")

# 1. Scatter Plot: Analyze relationship between Sepal Length and Sepal Width
plt.figure(figsize=(7, 4))
sns.scatterplot(data=iris_df, x='sepal_length', y='sepal_width', hue='species', palette='deep')
plt.title('Sepal Length vs Sepal Width across Species')
plt.xlabel('Sepal Length (cm)')
plt.ylabel('Sepal Width (cm)')
plt.tight_layout()
plt.show()

# 2. Histogram: Examine data distribution of Petal Length
plt.figure(figsize=(7, 4))
sns.histplot(data=iris_df, x='petal_length', hue='species', kde=True, bins=20, element='step')
plt.title('Distribution of Petal Length')
plt.xlabel('Petal Length (cm)')
plt.ylabel('Count')
plt.tight_layout()
plt.show()

# 3. Box Plot: Detect outliers and spread of Petal Width (Fixed to remove the warning)
plt.figure(figsize=(7, 4))
sns.boxplot(data=iris_df, x='species', y='petal_width', hue='species', palette='muted', legend=False)
plt.title('Petal Width Spread and Outliers by Species')
plt.xlabel('Species')
plt.ylabel('Petal Width (cm)')
plt.tight_layout()
plt.show()
```

```python
# --------------------------------------------------------------------
# CONCLUSION & INSIGHTS
# --------------------------------------------------------------------
print("  CONCLUSION SUMMARY ")
print("1. The Iris dataset contains 150 rows across 3 species with zero missing values.")
print("2. Setosa forms a highly distinct cluster with significantly smaller petal attributes.")
print("3. Box plots identify minor outliers in Sepal Width for the Virginica species, but overall data spread is highly consistent.")
```
